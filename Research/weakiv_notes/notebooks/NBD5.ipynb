```
import shutil, traceback, time
from spectraliv import __version__ as pkg_version
from spectraliv.experiments import (
    run_size_cell, run_power_cell, run_decisive_cell, run_robust_cell,
    run_scaling,
)

OUT_ROOT = "/content/results"
MASTER_SEED = 20260823

B_PARAMS = {
    "phase3_size_grid": 20000,
    "phase3_power_surface": 300,
    "phase3_decisive_grid": 400,
    "phase3_robustness": 4000,
}
PATCH_PARAMS = {"patch_reps": 250, "b_boot": 99}

CELLS = [
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.1",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.18",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.28",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.28,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.4",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.4,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.55",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.55,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p5_th0.72",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 600,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.72,
   "cell_id": "a0.3_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.05,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.1,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.18,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.28,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.4,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.55",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.55,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k0.5_none_p1_th0.72",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 0.5,
   "theta": 0.72,
   "cell_id": "a0.7_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.05,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.28,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.4,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.7_k2.0_none_p1_th0.72",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 699,
   "alpha": 0.7,
   "kappa": 2.0,
   "theta": 0.72,
   "cell_id": "a0.7_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.05",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.05,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.1",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.1,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.18",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.18,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.28",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.28,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.4",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.4,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.55",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.55,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p5_th0.72",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.72,
   "cell_id": "a0.1_k0.5_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.05",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.05,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.1",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.18",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.28",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.28,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.4",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.4,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.55",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.55,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p5_th0.72",
  "cell": {
   "n": 2000,
   "p": 5,
   "q": 200,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.72,
   "cell_id": "a0.1_k2.0_none_p5"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.05,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.1,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.18,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.28,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.4,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.55",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.55,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k0.5_none_p1_th0.72",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 0.5,
   "theta": 0.72,
   "cell_id": "a0.5_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k2.0_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 2.0,
   "theta": 0.05,
   "cell_id": "a0.5_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k2.0_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.5_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.5_k2.0_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 500,
   "alpha": 0.5,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.5_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.05,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.1,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.18,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.28,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.4,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k0.5_none_p1_th0.72",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 0.5,
   "theta": 0.72,
   "cell_id": "a0.3_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.05,
   "cell_id": "a0.3_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.3_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.3_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.3_k2.0_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 300,
   "alpha": 0.3,
   "kappa": 2.0,
   "theta": 0.28,
   "cell_id": "a0.3_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.05,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.1,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.18,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.28,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.4,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.55",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.55,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k0.5_none_p1_th0.72",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 0.5,
   "theta": 0.72,
   "cell_id": "a0.1_k0.5_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.05",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.05,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.1",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.1,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.18",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.18,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.28",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.28,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.4",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.4,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 },
 {
  "experiment": "phase3_decisive_grid",
  "cell_id": "a0.1_k2.0_none_p1_th0.55",
  "cell": {
   "n": 1000,
   "p": 1,
   "q": 100,
   "alpha": 0.1,
   "kappa": 2.0,
   "theta": 0.55,
   "cell_id": "a0.1_k2.0_none_p1"
  }
 }
]

def done_markers(exp, cid):
    d = os.path.join(OUT_ROOT, exp, "cells")
    if exp == "phase3_decisive_grid":
        names = ["_done_%s_coverage.csv.json" % cid,
                 "_done_%s_risk.csv.json" % cid]
    else:
        names = ["_done_%s.csv.json" % cid]
    return all(os.path.exists(os.path.join(d, nm)) for nm in names)


def checkpoint_download(tag):
    """Zip everything so far and push it to the browser. A dead session then
    only loses the cell that was in flight, never finished ones."""
    try:
        zpath = "/content/results_%s_ckpt_%s" % (NB_ID, tag)
        shutil.make_archive(zpath, "zip", OUT_ROOT)
        from google.colab import files
        files.download(zpath + ".zip")
        print("[ckpt] downloaded:", zpath + ".zip")
    except Exception as e:
        print("(checkpoint download skipped):", e)


summary = []
for item in CELLS:
    exp, cid = item["experiment"], item["cell_id"]
    if done_markers(exp, cid):
        print("[skip] %s (done markers present)" % cid)
        continue
    t0 = time.time()
    try:
        cell = dict(item["cell"])
        if exp == "phase3_size_grid":
            run_size_cell(cell, big_b=B_PARAMS[exp], b_cal_cv=4000,
                          out_root=OUT_ROOT)
        elif exp == "phase3_power_surface":
            run_power_cell(cell, reps_per_theta=B_PARAMS[exp], b_cal_cv=4000,
                           out_root=OUT_ROOT)
        elif exp == "phase3_decisive_grid":
            run_decisive_cell(cell, reps=B_PARAMS[exp], b_cal_cv=4000,
                              out_root=OUT_ROOT)
        elif exp == "phase3_robustness":
            run_robust_cell(cell, big_b=B_PARAMS[exp],
                            patch_reps=PATCH_PARAMS["patch_reps"],
                            b_boot=PATCH_PARAMS["b_boot"], out_root=OUT_ROOT)
        elif exp == "phase3_scaling":
            run_scaling(out_root=OUT_ROOT)
        print("[done] %s in %.1fs" % (cid, time.time() - t0))
        summary.append([cid, round(time.time() - t0, 1)])

        if exp == "phase3_decisive_grid":
            base = cell["cell_id"]; suf = "_th%s" % cell["theta"]
            cdir = os.path.join(OUT_ROOT, exp, "cells")
            for s in ("_coverage.csv", "_risk.csv", "_raw.npz"):
                src = os.path.join(cdir, base + s)
                if os.path.exists(src):
                    shutil.copyfile(src, os.path.join(cdir, base + suf + s))
                    if s != "_raw.npz":
                        dm = os.path.join(cdir, "_done_" + base + s + ".json")
                        dn = os.path.join(cdir, "_done_" + base + suf + s
                                          + ".json")
                        if os.path.exists(dm):
                            shutil.copyfile(dm, dn)
        checkpoint_download(cid.replace(".", ""))
    except Exception:
        print("[FAIL] %s" % cid)
        traceback.print_exc()

manifest = {
    "notebook_id": NB_ID, "git_sha": GIT_SHA,
    "package_version": pkg_version, "master_seed": MASTER_SEED,
    "b_params": B_PARAMS, "patch_params": PATCH_PARAMS,
    "cells": [c["cell_id"] for c in CELLS],
    "timings_s": summary,
}
with open(os.path.join(OUT_ROOT, "manifest_%s.json" % NB_ID), "w") as f:
    json.dump(manifest, f, indent=1)
print(json.dumps(manifest, indent=1))
```
